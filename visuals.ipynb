```python
import torch as t
import argparse
import pandas as pd
from tqdm import tqdm
import os
import configparser
from nnsight import LanguageModel
```

```python
import json

json_path = "/workspace/geometry-of-truth/experimental_outputs/few_shot_results.json"

models_to_show = [
    "llama-3.1-8B",
    "llama-3.1-8B-Instruct",
]

# --------------------
# LOAD JSON
# --------------------
with open(json_path, "r") as f:
    data = json.load(f)

df = pd.DataFrame(data)

# --------------------
# FILTER MODELS
# --------------------
df = df[df["model"].isin(models_to_show)]

# --------------------
# BUILD & PRINT TABLES
# --------------------
for model in models_to_show:
    df_model = df[df["model"] == model]

    # one row per dataset
    table = (
        df_model
        .groupby("dataset")["acc"]
        .mean()
        .sort_index()
        .to_frame(name="accuracy")
    )

    print("\n" + "=" * 50)
    print(f"Model: {model}")
    print("=" * 50)
    print(table.round(3))
```

```python
for model_name in ['llama-3.1-8B', 'llama-3.1-8B-Instruct']:
    df = pd.read_csv(f"/workspace/geometry-of-truth/reponses/{model_name}/repsonses.csv")
    df["response"] = df["response"].str.replace(" ", "", regex=False).str.lower()
    df["pred"] = df["response"].map({"true": 1, "false": 0})
    df["pred"] = df["response"].map({"true": 1, "false": 0})
    df["correct"] = (df["pred"] == df["label"])
    accuracy_per_dataset = df.groupby("dataset")["correct"].mean()
    print(model_name)
    print(accuracy_per_dataset)
```
