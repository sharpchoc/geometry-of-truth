```python
import json
import pandas as pd
import numbers
```

```python
# load json
with open("experimental_outputs/label_change_intervention_results.json", "r") as f:
    data = json.load(f)

# filter entries with numeric intervention_strength

rows = []
for d in data:
    if "intervention_strength" in d and isinstance(d["intervention_strength"], numbers.Number):
        rows.append({
            "model": d["model"],
            "probe": d["probe class"],
            "intervention_strength": d["intervention_strength"],
            "p_diff": d["p_diff"],
        })

df = pd.DataFrame(rows)

# create pivot table
table = df.pivot_table(
    index=["model", "probe"],
    columns="intervention_strength",
    values="p_diff",
    aggfunc="mean"   # safe if duplicates exist
)

print(table)
```
